```python
x = data.drop('Class',axis=1)
y = data['Class']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x['Amount'] = scaler.fit_transform(x[['Amount']])
x['Time'] = scaler.fit_transform(x[['Time']])
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test , y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
print(y_train.value_counts)
print(y_test.value_counts)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

model = LogisticRegression()
model.fit(x_train,y_train)

y_pred = model.predict(x_test)

print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE()
x_train_smote, y_train_smote = smote.fit_resample(x_train,y_train)

print(y_train_smote.value_counts())
```

```python
# SMOTE Ke Baad Train Karo
lr_smote = LogisticRegression()
lr_smote.fit(x_train_smote, y_train_smote)

# Predict Karo
y_pred_smote = lr_smote.predict(x_test)

# Results Dekho
print(classification_report(y_test, y_pred_smote))
print(confusion_matrix(y_test, y_pred_smote))
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    scale_pos_weight=577,
    random_state=42
)

xgb.fit(x_train, y_train)

y_pred_xgb = xgb.predict(x_test)

print(classification_report(y_test, y_pred_xgb))
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import f1_score

# Training Pe Score
y_train_pred = xgb.predict(x_train)
train_f1 = f1_score(y_train, y_train_pred)

y_test_pred = xgb.predict(x_test)
test_f1 = f1_score(y_test, y_test_pred)

print(f"Training F1: {train_f1}")
print(f"Testing F1:  {test_f1}")
```

```python
from sklearn.metrics import f1_score, recall_score

y_train_pred = xgb.predict(x_train)
train_f1 = f1_score(y_train, y_train_pred)
train_recall = recall_score(y_train, y_train_pred)

test_f1 = f1_score(y_test, y_pred_xgb)
test_recall = recall_score(y_test, y_pred_xgb)

print(f"Training F1:     {train_f1:.2f}")
print(f"Testing F1:      {test_f1:.2f}")
print(f"Training Recall: {train_recall:.2f}")
print(f"Testing Recall:  {test_recall:.2f}")
```

```python
import shap

# XGBoost Ke Liye TreeExplainer
explainer = shap.TreeExplainer(xgb)

# SHAP Values Calculate Karo
shap_values = explainer.shap_values(x_test)

print("SHAP Values Shape:", shap_values.shape)
```

```python
import matplotlib.pyplot as plt

shap.summary_plot(
    shap_values,
    x_test,
    plot_type="bar"
)
plt.show()
```

```python
shap.summary_plot(
    shap_values,
    x_test
)
plt.show()
```

```python
fraud_indices = y_test[y_test == 1].index
first_fraud = fraud_indices[0]

idx = x_test.index.get_loc(first_fraud)

shap.initjs()
shap.force_plot(
    explainer.expected_value,
    shap_values[idx],
    x_test.iloc[idx],
    matplotlib=True
)
```

```python
# Fraud Transaction Ka Index
fraud_indices = y_test[y_test == 1].index
first_fraud = fraud_indices[0]
idx = x_test.index.get_loc(first_fraud)

# Waterfall Plot (Zyada Clear Hota Hai)
shap.waterfall_plot(
    shap.Explanation(
        values=shap_values[idx],
        base_values=explainer.expected_value,
        data=x_test.iloc[idx],
        feature_names=x_test.columns.tolist()
    )
)
```

```python
# Pehle Colab Me Jaa Aur Yeh Run Kar:

# 5 Normal Aur 5 Fraud Transactions Nikaalo
import json

# 5 Fraud Samples
fraud_samples = x_test[y_test == 1].head(5)
# 5 Normal Samples
normal_samples = x_test[y_test == 0].head(5)

# Combine
samples = []

for i, (idx, row) in enumerate(fraud_samples.iterrows()):
    samples.append({
        "id": f"fraud_{i+1}",
        "type": "Fraud Transaction",
        "data": row.to_dict()
    })

for i, (idx, row) in enumerate(normal_samples.iterrows()):
    samples.append({
        "id": f"normal_{i+1}",
        "type": "Normal Transaction",
        "data": row.to_dict()
    })

# Save As JSON
with open('samples.json', 'w') as f:
    json.dump(samples, f, indent=2)

print("Samples Saved!")
```
